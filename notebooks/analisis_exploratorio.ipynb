- Títulos de juego duplicados (equivalente a reviews por juego)

```python
print("Los top 30 comentarios más populares son:")
result = df.groupby('review').size().sort_values(ascending=False).head(30)
result
```

De los 30 comentarios más populares 29 son positivos y solo 1 es neutral (el punto), con varios de ellos que son similares pero cambian por errores ortográficos. A su vez, hay caracteres que a criterio nuestro son positivos, así como 👍, W, 8/10 (que se clasifica como promotor), gud, peak.

```python
import numpy as np
import matplotlib.pyplot as plt

values = df["author.num_reviews"].dropna()

plt.hist(np.log1p(values), bins=30, edgecolor="black")
plt.xlabel("log(1 + number of reviews)")
plt.ylabel("Number of authors")
plt.title("Distribution of author reviews (log-transformed)")
plt.show()
```
