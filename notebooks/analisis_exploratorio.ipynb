## Resumen del EDA

```python
missing = (
    df.isna().sum()
      .to_frame("missing")
      .assign(percent=lambda x: 100 * x["missing"] / len(df))
      .sort_values("percent", ascending=False)
)
top_missing = missing[missing["missing"] > 0].head(20).sort_values("percent")
if not top_missing.empty:
    ax = top_missing["percent"].plot(kind="barh", figsize=(9, 6))
    ax.set_xlabel("Missing values (%)")
    ax.set_title("Columnas con más NA's")
    plt.tight_layout()
    plt.show()
```

#### Largo de las reviews

```python
reviews = df["review"].fillna("").astype(str)
review_chars = reviews.str.len()
review_words = reviews.str.split().str.len()

plt.hist(np.log1p(review_words), bins=40, edgecolor="black")
plt.xlabel("log(1 + number of words)")
plt.ylabel("Number of reviews")
plt.title("Review length distribution")
plt.tight_layout()
plt.show()
```

### Outliers de las variables numéricas

```python

count_candidates = [
    "votes_up", "votes_funny", "comment_count",
    "author.num_games_owned", "author.num_reviews",
    "author.playtime_forever", "author.playtime_last_two_weeks",
    "author.playtime_at_review", "author.deck_playtime_at_review",
    "hardware.system_ram", "hardware.vram_size",
]
count_cols = [col for col in count_candidates if col in df.columns]
counts = df[count_cols].apply(pd.to_numeric, errors="coerce")

outlier_rows = []
for col in count_cols:
    values = counts[col].dropna()
    if values.empty:
        continue

    q1, q3 = values.quantile([0.25, 0.75])
    iqr = q3 - q1
    lower, upper = q1 - 1.5 * iqr, q3 + 1.5 * iqr
    outlier_rows.append({
        "column": col,
        "outlier_count": int(((values < lower) | (values > upper)).sum()),
        "outlier_percent": 100 * ((values < lower) | (values > upper)).mean(),
        "upper_iqr_limit": upper,
    })

display(pd.DataFrame(outlier_rows).set_index("column"))



```

```python
plot_cols = [
    col for col in count_cols
    if counts[col].notna().any() and (counts[col].dropna() >= 0).all()
]
if plot_cols:
    box_data = [np.log1p(counts[col].dropna()) for col in plot_cols]
    plt.figure(figsize=(12, 6))
    plt.boxplot(box_data, tick_labels=plot_cols, showfliers=True)
    plt.xticks(rotation=65, ha="right")
    plt.ylabel("log(1 + value)")
    plt.title("Potential outliers in review and author activity")
    plt.tight_layout()
    plt.show()
```

### Correlaciones entre variables

```python

corr = counts.corr()
if corr.shape[0] >= 2:
    fig, ax = plt.subplots(figsize=(9, 7))
    image = ax.imshow(corr, vmin=-1, vmax=1, cmap="coolwarm")
    ax.set_xticks(range(len(corr.columns)), corr.columns, rotation=75, ha="right")
    ax.set_yticks(range(len(corr.index)), corr.index)
    ax.set_title("Correlatción entre variables candidatas")
    fig.colorbar(image, ax=ax, label="Correlation")
    plt.tight_layout()
    plt.show()
```
